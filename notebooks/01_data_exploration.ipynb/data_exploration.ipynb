```python
#to avoide repeating the same code for each dataframe we can create a function to profile the dataset(Resuable profiling function)

def profile_dataset(df, name):

    print(f"\n========== {name} ==========")

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nNull Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nSample Data:")
    print(df.head())
```

```python
profile_dataset(trials,"Trials")
```

```python
profile_dataset(patients, "Patients")
```

```python
profile_dataset(sites, "Sites")
```

```python
profile_dataset(visits, "Visits")
```

# Dataset Relationships

## trials → patients
Connected using: trial_id

One trial can have many patients.

---

## trials → sites
Connected using: trial_id

One trial can run across multiple sites.

---

## patients → visits
Connected using: patient_id

One patient can have multiple visits.

---

## patients → adverse_events
Connected using: patient_id

One patient can report multiple adverse events.

---

## patients → protocol_deviations
Connected using: patient_id

One patient can have multiple protocol deviations.

---

## sites → drug_inventory
Connected using: site_id

Each site maintains drug inventory records.

---

## sites → drug_shipments
Connected using: site_id

Drug shipments are sent to sites.

---

## visits → lab_results
Connected using: visit_id

Lab results are recorded during patient visits.
